```python
#DataFrame
#using dictionary to list
info={
    "Name": ["a", "B", "C"],
    "Age": [21, 22, 23],
     "GPA": [8, 8.5, 9]
}
df=pd.DataFrame(info)
print(df)
```

```python
df
```

```python
#using list to list'
df=pd.DataFrame([[1,2],[5,6]], columns=["Nmae","Age"])
print(df)
```

```python
#numpy array to list
import numpy as np
mp_arr=np.array([[1,2,3],[4,5,6]])
df=pd.DataFrame(mp_arr, columns=["A", "B", "C"])
print(df)
```
