```python
import requests
```

```python
URL="https://www.scrapethissite.com/pages/simple/"
res=requests.get(URL)
```

```python
print(res.status_code)
#if res.status_code ==200:
    #print(res.text) ## prints HTML content
    #print(res.content) ## prints raw byte response
    #print(res.headers) ## prints metadata

with open("Scrapped_data/data1.html", "w",encoding="utf-8") as f:
    f.write(res.text)
```

```python
from bs4 import BeautifulSoup
```

```python
with open("Scrapped_data/data1.html", "r") as f:
    html_content=f.read()

soup=BeautifulSoup(html_content,"lxml")
```

```python
soup
soup.find("h1")
all_h3=soup.find_all("h3")
all_countries=[]
for h3 in all_h3:
    name=h3.get_text(strip=True) #get.text extracts the name of country from the html tag and strip() usees to remove the spaces

    population=(h3.find_next("div").select("span.country-population")[0].get_text(strip=True))

    all_countries.append([name,population])

print(all_countries)
```
