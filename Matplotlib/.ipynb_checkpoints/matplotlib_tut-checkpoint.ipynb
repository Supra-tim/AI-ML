```python
#real data

#list of oscar movies

oscar_movies=[
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "Argo"
]

years=[2008, 2009, 2010, 2011, 2012]
oscar_revenue=[1005, 170, 427, 133, 232]

plt.plot(years,oscar_revenue)
plt.title("Oscar Movies in year vs revenue")
plt.xlabel("years")
plt.ylabel("revenue")
```

```python
#non - oscar winner movies

non_oscar_movies=["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Linclon"]
non_oscar_revenue=[378,2788,829, 185,275]

plt.plot(years, oscar_revenue,"o--r", label="Oscar movies")
plt.plot(years, non_oscar_revenue,"<-y", label="Non pscar Movies")

plt.title("Oscar vs Non Oscar movie")

plt.xlabel("years")
plt.ylabel("revenue")
plt.legend()
```

```python
with plt.xkcd():
    plt.plot(years, oscar_revenue, label="Oscar movies")
    plt.plot(years, non_oscar_revenue, label="Non pscar Movies")
    
    plt.title("Oscar vs Non Oscar movie")
    
    plt.xlabel("years")
    plt.ylabel("revenue")
    plt.legend()
    
    ##plt.style.available
    ##plt.style.use("default")
    ##plt.grid()
```

```python
## bar plot

oscar_movies=[
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "Argo"
]

years=[2008, 2009, 2010, 2011, 2012]
oscar_revenue=[1005, 170, 427, 133, 232]
plt.bar(years, oscar_revenue)
plt.title("Years vs revenue")
plt.xlabel("years")
plt.ylabel("revenue")

for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+20, str(oscar_revenue[i]), ha="center")

plt.ylim(0, max(oscar_revenue)+100)
```

```python
import numpy as np
oscar_movies=[
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "Argo"
]

oscar_revenue=[1005, 170, 427, 133, 232]

non_oscar_movies=["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Linclon"]
non_oscar_revenue=[378,2788,829, 185,275]

years=[2008, 2009, 2010, 2011, 2012]

x=np.arange(len(years))
width=0.4

plt.bar(x-width/2,  oscar_revenue, width, label="oscar Movies")
plt.bar(x+width/2,  non_oscar_revenue, width, label="non- oscar movies")
plt.title("bi graph")
plt.xlabel("Years")
plt.ylabel("revenue")
plt.legend()
plt.xticks(x,years)
```

```python
# Horizontal 

oscar_movies=[
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "Argo"
]

oscar_revenue=[1005, 170, 427, 133]
plt.barh(oscar_movies,oscar_revenue)
plt.xlabel("revenue")
plt.ylabel("Movies")
plt.title("movies vs revenue")
```

```python
#scatter plots

people=["a", "b", "c", "d", "e","f", "g", "h", "i", "j"]
age=[22,25,30,35,40,45,50,55,60,65]
blood_presure=[110, 115, 120, 122, 135, 140, 123, 150, 115, 165]

colors=["green" if x <130 else "red" for x in blood_presure]
##plt.scatter(age, blood_presure, color=colors, s=blood_presure, marker="o", alpha=0.5)
plt.scatter(age, blood_presure, s=blood_presure, cmap="OrRd", c=blood_presure) 
plt.xlabel("age")
plt.ylabel("revenue")
plt.title("presure")
plt.grid()
plt.colorbar()

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], blood_presure[i]), xytext=(age[i]+1, blood_presure[i]+1))

plt.xlim(min(age), max(age)+10)
plt.ylim(min(blood_presure), max(blood_presure)+10)
```

```python
#multiple data in Scatter plots

cities=["a", "b", "c", "d", "e"]

#winter temperature vs humidity 
winter_temp=[5,2,6,3,1]
winter_humidity=[80, 60, 45, 90, 10]

#summer temperature vs humidity
summer_temp=[30, 45, 54, 25, 22]
summer_humidity=[60, 50, 57, 23, 45]

plt.scatter(winter_temp, winter_humidity , label="winter")
plt.scatter(summer_temp, summer_humidity, label="Summer")

plt.xlabel("temperature")
plt.ylabel("Humidity")
plt.title("Temp vs Humidity")
plt.legend()
```

```python
# pie charts

expences=["salaries", "rent", "r&d", "marketing", "mislennious"]
amounts=[500, 150, 120, 100, 50]
explodes=[0,0,0.2, 0,0]
plt.pie(amounts, labels=expences, autopct="%1.1f%%", 
        wedgeprops ={ "edgecolor":"black",
                    "linewidth":1,
                    "linestyle":"--",},
        explode=explodes,
        startangle=180,
        shadow=True
       )
```

```python
# Histogram

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
scores=np.random.normal(70,10,100) #normal distribution

bins=[0,30,40,50,70,80,90,100]
plt.hist(scores, edgecolor="black", bins=bins)
plt.xlabel("scores")
plt.ylabel("# of sttudents")
print(scores)
```
