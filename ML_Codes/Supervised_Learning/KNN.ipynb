```python
import pandas as pd
from sklearn.metrics import precision_score, accuracy_score, recall_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
heart_df=pd.read_csv("heart.csv")
X=heart_df.drop("target", axis=1)
Y=heart_df["target"]
X_tarin, X_test, y_train, y_test=train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
scaler=StandardScaler()
X_tarin_scaled= scaler.fit_transform(X_tarin)
X_test_scaled=scaler.transform(X_test)
```

```python
knn_classification=KNeighborsClassifier(n_neighbors=3)
knn_classification.fit(X_tarin_scaled, y_train)
y_pred=knn_classification.predict(X_test_scaled)
print("Recall_Score:", recall_score(y_test, y_pred))
print("Accuracy score:", accuracy_score(y_test, y_pred))
print("precidion score:", precision_score(y_test, y_pred))
```

```python
knn_classification=KNeighborsClassifier(n_neighbors=5)
knn_classification.fit(X_tarin_scaled, y_train)
y_pred=knn_classification.predict(X_test_scaled)
print("Recall_Score:", recall_score(y_test, y_pred))
print("Accuracy score:", accuracy_score(y_test, y_pred))
print("precidion score:", precision_score(y_test, y_pred))
```
