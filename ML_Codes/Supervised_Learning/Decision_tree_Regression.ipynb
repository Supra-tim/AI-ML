```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
```

```python
df=load_diabetes(as_frame=True).frame
```

```python
x=df.drop("target", axis=1)
y=df["target"]
```

```python
x.head()
```

```python
X_train, X_test, y_train, y_test=train_test_split(x,y, test_size=0.3, random_state=42)
```

```python
from sklearn.tree import DecisionTreeRegressor
model=DecisionTreeRegressor(max_depth=7, min_samples_split=20)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import r2_score, mean_squared_error
y_pred_train=model.predict(X_train)
y_pred_test=model.predict(X_test)
print("MSE tarin:" , mean_squared_error(y_train, y_pred_train))
print("MSE test:" , mean_squared_error(y_test, y_pred_test))

print("r^2 tarin:" , r2_score(y_train, y_pred_train))
print("r^2 test:" , r2_score(y_test, y_pred_test))
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(18,18))
plot_tree(
    model,
    feature_names=x.columns,
    #max_depth=2,
    filled=True
)
plt.tight_layout()
plt.show()
```
