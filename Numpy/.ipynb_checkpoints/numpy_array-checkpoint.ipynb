```python
import numpy as np
```

```python
arr=np.array([1,2,3,4])
print(arr, type(arr),arr.shape)

arr2=np.array([1,2,3,"Hello"])
print(arr2, type(arr2), arr.shape)

arr2D=np.array([[1,2,3],[4,5,6]])
print(arr2D, type(arr2D), arr2D.shape)
```

```python
#creating arry using built in function
arr1=np.zeros((2,3))
print(arr1, arr1.shape)

arr2=np.ones((5,))
print(arr2,arr2.shape)

arr3=np.full((2,3),100)
print(arr3, arr3.shape)

arr4=np.eye(3)
print(arr4, arr4.shape)

arr5=np.arange(1,10)
print(arr5, arr5.shape)

arr6=np.linspace(1,100,4)
print(arr6, arr6.shape)
```

```python
#opersations in Numpy
#reshaping

arr=np.array([[1,2,3],[4,5,6]])
print(arr,arr.shape)

resized=arr.reshape((3.2))
print(resized,resized.shape)
```
